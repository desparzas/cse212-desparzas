This notebook provides a comprehensive guide to understanding and implementing recursive solutions for the W05 Code assignment. We'll explore each problem step by step, demonstrating the recursive thinking process and implementation strategies.

## Key Recursion Principles
1. **Base Case**: A condition that stops the recursion
2. **Smaller Problem**: Each recursive call should work on a reduced version of the original problem
3. **Recursive Case**: The function calls itself with modified parameters
4. **Memoization**: Caching results to improve performance when needed

## Problem 1: Recursive Squares Sum

**Problem**: Find the sum 1² + 2² + 3² + ... + n² using recursion.

### Recursive Thinking Process:
1. **What's the smaller problem?** SumSquares(n-1)
2. **How does the current problem relate to the smaller one?** SumSquares(n) = n² + SumSquares(n-1)
3. **What's the base case?** When n ≤ 0, return 0

Let's implement this in Python to understand the concept:

````python
def sum_squares_recursive(n):
    """
    Recursively calculate the sum of squares from 1 to n
    Formula: 1² + 2² + 3² + ... + n²
    """
    # Base case: if n <= 0, return 0
    if n <= 0:
        return 0
    else:
        # Recursive case: n² + sum of squares from 1 to (n-1)
        return (n * n) + sum_squares_recursive(n - 1)

# Test the function
test_values = [1, 2, 3, 4, 5, 10]
print("Sum of Squares Results:")
for n in test_values:
    result = sum_squares_recursive(n)
    print(f"SumSquares({n}) = {result}")

# Verify with mathematical formula: n(n+1)(2n+1)/6
print("\nVerification with formula n(n+1)(2n+1)/6:")
for n in test_values:
    formula_result = n * (n + 1) * (2 * n + 1) // 6
    recursive_result = sum_squares_recursive(n)
    print(f"n={n}: Formula={formula_result}, Recursive={recursive_result}, Match={formula_result == recursive_result}")
````

## Problem 2: Permutations Choose

**Problem**: Generate all permutations of length `size` from a list of unique letters.

### Recursive Thinking Process:
1. **Base case**: When the current word length equals the desired size, add it to results
2. **Recursive case**: For each available letter, add it to the current word and recursively find permutations with remaining letters
3. **Smaller problem**: Remove one letter from available letters and increase word length by 1

Let's visualize this with a tree diagram first:

````python
def permutations_choose(results, letters, size, word=""):
    """
    Generate all permutations of length 'size' from 'letters'
    """
    # Base case: if we've built a word of the desired size
    if len(word) == size:
        results.append(word)
        return
    
    # Recursive case: try each available letter
    for i in range(len(letters)):
        # Take current letter and remove it from available letters
        current_letter = letters[i]
        remaining_letters = letters[:i] + letters[i+1:]
        
        # Recursively build permutations with this letter added
        permutations_choose(results, remaining_letters, size, word + current_letter)

# Test with different examples
test_cases = [
    ("ABC", 2),
    ("ABCD", 3),
    ("XYZ", 1)
]

for letters, size in test_cases:
    results = []
    permutations_choose(results, letters, size)
    print(f"Permutations of length {size} from '{letters}':")
    print(f"Results: {sorted(results)}")
    print(f"Count: {len(results)} (Expected: {len(letters) * (len(letters)-1) if size == 2 else 'varies'})\n")
````

## Problem 3: Climbing Stairs with Memoization

**Problem**: Count ways to climb `s` stairs when you can take 1, 2, or 3 steps at a time.

### Recursive Thinking:
- **Formula**: CountWays(s) = CountWays(s-1) + CountWays(s-2) + CountWays(s-3)
- **Base cases**: s=0→0, s=1→1, s=2→2, s=3→4 ways
- **Memoization**: Store computed results to avoid redundant calculations

Let's first see the performance difference between naive recursion and memoization:

````python
def count_ways_naive(s):
    """Naive recursive implementation (slow for large s)"""
    if s == 0: return 0
    if s == 1: return 1
    if s == 2: return 2
    if s == 3: return 4
    
    return count_ways_naive(s-1) + count_ways_naive(s-2) + count_ways_naive(s-3)

def count_ways_memoized(s, memo=None):
    """Memoized recursive implementation (fast)"""
    if memo is None:
        memo = {}
    
    # Base cases
    if s == 0: return 0
    if s == 1: return 1
    if s == 2: return 2
    if s == 3: return 4
    
    # Check if already computed
    if s in memo:
        return memo[s]
    
    # Compute and store result
    result = count_ways_memoized(s-1, memo) + count_ways_memoized(s-2, memo) + count_ways_memoized(s-3, memo)
    memo[s] = result
    return result

# Test small values with both approaches
small_values = [1, 2, 3, 4, 5, 10, 15]
print("Climbing Stairs Results (small values):")
for s in small_values:
    result = count_ways_memoized(s)
    print(f"CountWays({s}) = {result}")

# Performance comparison
print("\nPerformance Comparison (timing in seconds):")
test_value = 20

# Time naive approach
start_time = time.time()
naive_result = count_ways_naive(test_value)
naive_time = time.time() - start_time

# Time memoized approach
start_time = time.time()
memo_result = count_ways_memoized(test_value)
memo_time = time.time() - start_time

print(f"CountWays({test_value}):")
print(f"  Naive: {naive_result} (Time: {naive_time:.4f}s)")
print(f"  Memoized: {memo_result} (Time: {memo_time:.6f}s)")
print(f"  Speedup: {naive_time/memo_time:.1f}x faster with memoization")
````

## Problem 4: Wildcard Binary Patterns

**Problem**: Generate all possible binary strings from a pattern containing wildcard '*' characters.

### Recursive Thinking:
1. **Base case**: If no wildcards remain, add the pattern to results
2. **Recursive case**: Find first wildcard, replace with '0' and '1', then recurse
3. **Smaller problem**: Each recursion has one fewer wildcard

Let's implement and visualize this:

````python
def wildcard_binary(pattern, results):
    """
    Generate all possible binary strings from a pattern with wildcards
    """
    # Base case: if no wildcards, add pattern to results
    wildcard_index = pattern.find('*')
    if wildcard_index == -1:
        results.append(pattern)
        return
    
    # Recursive case: replace first wildcard with '0' and '1'
    # Replace with '0'
    pattern_with_0 = pattern[:wildcard_index] + '0' + pattern[wildcard_index + 1:]
    wildcard_binary(pattern_with_0, results)
    
    # Replace with '1'
    pattern_with_1 = pattern[:wildcard_index] + '1' + pattern[wildcard_index + 1:]
    wildcard_binary(pattern_with_1, results)

# Test with various patterns
test_patterns = [
    "1*0",      # 2 results
    "**1",      # 4 results
    "10*0*",    # 4 results
    "***",      # 8 results
    "1010"      # 1 result (no wildcards)
]

for pattern in test_patterns:
    results = []
    wildcard_binary(pattern, results)
    results.sort()
    
    wildcards = pattern.count('*')
    expected = 2 ** wildcards
    
    print(f"Pattern: '{pattern}' (wildcards: {wildcards})")
    print(f"Results: {results}")
    print(f"Count: {len(results)} (Expected: {expected})\n")
````

### Visualization of the Binary Tree Process:

````python
# Let's trace through the process for pattern "1**"
def trace_wildcard_binary(pattern, results, depth=0, trace_list=None):
    """Trace the recursive process for educational purposes"""
    if trace_list is None:
        trace_list = []
    
    indent = "  " * depth
    trace_list.append(f"{indent}Processing: '{pattern}'")
    
    wildcard_index = pattern.find('*')
    if wildcard_index == -1:
        results.append(pattern)
        trace_list.append(f"{indent}-> Added to results: '{pattern}'")
        return trace_list
    
    # Replace with '0'
    pattern_with_0 = pattern[:wildcard_index] + '0' + pattern[wildcard_index + 1:]
    trace_list.append(f"{indent}-> Trying with '0': '{pattern_with_0}'")
    trace_wildcard_binary(pattern_with_0, results, depth + 1, trace_list)
    
    # Replace with '1'
    pattern_with_1 = pattern[:wildcard_index] + '1' + pattern[wildcard_index + 1:]
    trace_list.append(f"{indent}-> Trying with '1': '{pattern_with_1}'")
    trace_wildcard_binary(pattern_with_1, results, depth + 1, trace_list)
    
    return trace_list

# Trace through "1**"
results = []
trace = trace_wildcard_binary("1**", results)

print("Trace of wildcard binary expansion for '1**':")
print("=" * 50)
for step in trace:
    print(step)

print(f"\nFinal results: {sorted(results)}")
````

## Problem 5: Maze Solver

**Problem**: Find all paths through a maze from (0,0) to the end position using recursion.

### Maze Representation:
- 0 = Wall (can't pass through)
- 1 = Open path (can move through)
- 2 = End goal (destination)

### Recursive Strategy:
1. **Base case**: If current position is the end, add path to results
2. **Recursive case**: Try moving in all 4 directions (up, down, left, right)
3. **Backtracking**: Add position when entering, remove when leaving
4. **Validation**: Use helper functions to check valid moves

Let's implement a simplified version:

````python
class SimpleMaze:
    def __init__(self, width, height, data):
        self.width = width
        self.height = height
        self.data = data
    
    def is_end(self, x, y):
        """Check if position is the end goal"""
        return self.data[y * self.height + x] == 2
    
    def is_valid_move(self, curr_path, x, y):
        """Check if move to (x,y) is valid"""
        # Check boundaries
        if x < 0 or x >= self.width or y < 0 or y >= self.height:
            return False
        
        # Check if it's a wall
        if self.data[y * self.height + x] == 0:
            return False
        
        # Check if already visited in current path
        if (x, y) in curr_path:
            return False
        
        return True
    
    def display(self, path=None):
        """Display the maze with optional path highlighted"""
        symbols = {0: '█', 1: '·', 2: 'E'}
        
        for y in range(self.height):
            row = ""
            for x in range(self.width):
                if path and (x, y) in path:
                    row += 'P' if (x, y) != (0, 0) and not self.is_end(x, y) else symbols[self.data[y * self.height + x]]
                else:
                    row += symbols[self.data[y * self.height + x]]
            print(row)

def solve_maze(results, maze, x=0, y=0, curr_path=None):
    """Recursively solve the maze"""
    if curr_path is None:
        curr_path = []
    
    # Add current position to path
    curr_path.append((x, y))
    
    # Base case: if we reached the end
    if maze.is_end(x, y):
        results.append(curr_path.copy())  # Add copy of path
        curr_path.pop()  # Backtrack
        return
    
    # Try all four directions: right, down, left, up
    directions = [(1, 0), (0, 1), (-1, 0), (0, -1)]
    direction_names = ['right', 'down', 'left', 'up']
    
    for (dx, dy), name in zip(directions, direction_names):
        new_x, new_y = x + dx, y + dy
        if maze.is_valid_move(curr_path, new_x, new_y):
            solve_maze(results, maze, new_x, new_y, curr_path)
    
    # Backtrack: remove current position
    curr_path.pop()

# Create a simple 3x3 test maze
# Layout:
# 1 1 1
# 1 0 1  (0 is wall)
# 1 1 2  (2 is end)
test_maze = SimpleMaze(3, 3, [1, 1, 1, 1, 0, 1, 1, 1, 2])

print("Test Maze Layout:")
test_maze.display()
print("Legend: █=Wall, ·=Path, E=End, P=Solution Path")

# Solve the maze
solutions = []
solve_maze(solutions, test_maze)

print(f"\nFound {len(solutions)} solution(s):")
for i, path in enumerate(solutions):
    print(f"\nSolution {i+1}: {path}")
    test_maze.display(path)
````

### Visualizing the Backtracking Process:

````python
def solve_maze_with_trace(results, maze, x=0, y=0, curr_path=None, depth=0, trace_list=None):
    """Solve maze with detailed tracing for educational purposes"""
    if curr_path is None:
        curr_path = []
    if trace_list is None:
        trace_list = []
    
    indent = "  " * depth
    trace_list.append(f"{indent}Visiting ({x}, {y})")
    
    # Add current position to path
    curr_path.append((x, y))
    trace_list.append(f"{indent}Current path: {curr_path}")
    
    # Base case: if we reached the end
    if maze.is_end(x, y):
        results.append(curr_path.copy())
        trace_list.append(f"{indent}*** SOLUTION FOUND! ***")
        curr_path.pop()
        trace_list.append(f"{indent}Backtracking from end")
        return trace_list
    
    # Try all four directions
    directions = [(1, 0), (0, 1), (-1, 0), (0, -1)]
    direction_names = ['right', 'down', 'left', 'up']
    
    for (dx, dy), name in zip(directions, direction_names):
        new_x, new_y = x + dx, y + dy
        if maze.is_valid_move(curr_path, new_x, new_y):
            trace_list.append(f"{indent}Trying to go {name} to ({new_x}, {new_y})")
            solve_maze_with_trace(results, maze, new_x, new_y, curr_path, depth + 1, trace_list)
        else:
            trace_list.append(f"{indent}Cannot go {name} to ({new_x}, {new_y}) - invalid move")
    
    # Backtrack
    curr_path.pop()
    trace_list.append(f"{indent}Backtracking from ({x}, {y})")
    
    return trace_list

# Trace through the maze solving process
results = []
trace = solve_maze_with_trace(results, test_maze)

print("Detailed trace of maze solving process:")
print("=" * 60)
for step in trace[:30]:  # Show first 30 steps to avoid too much output
    print(step)

if len(trace) > 30:
    print(f"... (showing first 30 of {len(trace)} total steps)")

print(f"\nFinal results: Found {len(results)} solution(s)")
````

### C# Implementation:
```csharp
public static void SolveMaze(List<string> results, Maze maze, int x = 0, int y = 0, List<ValueTuple<int, int>>? currPath = null)
{
    // Initialize path if first call
    if (currPath == null) {
        currPath = new List<ValueTuple<int, int>>();
    }
    
    // Add current position to path
    currPath.Add((x, y));

    // Base case: if we reached the end
    if (maze.IsEnd(x, y))
    {
        results.Add(currPath.AsString());
        currPath.RemoveAt(currPath.Count - 1);  // Backtrack
        return;
    }

    // Try all four directions: right, down, left, up
    if (maze.IsValidMove(currPath, x + 1, y))  // right
        SolveMaze(results, maze, x + 1, y, currPath);
    
    if (maze.IsValidMove(currPath, x, y + 1))  // down
        SolveMaze(results, maze, x, y + 1, currPath);
    
    if (maze.IsValidMove(currPath, x - 1, y))  // left
        SolveMaze(results, maze, x - 1, y, currPath);
    
    if (maze.IsValidMove(currPath, x, y - 1))  // up
        SolveMaze(results, maze, x, y - 1, currPath);

    // Backtrack: remove current position
    currPath.RemoveAt(currPath.Count - 1);
}
```

````python
def analyze_complexity():
    """Analyze and visualize the complexity of our recursive solutions"""
    print("TIME COMPLEXITY ANALYSIS")
    print("=" * 50)
    
    algorithms = [
        ("Sum of Squares", "O(n)", "Linear - each number processed once"),
        ("Permutations Choose", "O(P(n,k))", "P(n,k) = n!/(n-k)! permutations generated"),
        ("Climbing Stairs (naive)", "O(3^n)", "Each step branches into 3 possibilities"),
        ("Climbing Stairs (memoized)", "O(n)", "Each subproblem solved only once"),
        ("Wildcard Binary", "O(2^w)", "w = number of wildcards, each branches to 2"),
        ("Maze Solver", "O(4^(m*n))", "Worst case: 4 directions at each of m*n cells")
    ]
    
    for name, complexity, explanation in algorithms:
        print(f"\n{name}:")
        print(f"  Complexity: {complexity}")
        print(f"  Explanation: {explanation}")
    
    # Create a visual comparison of growth rates
    n_values = np.arange(1, 11)
    
    plt.figure(figsize=(12, 8))
    
    # Plot different complexity curves
    plt.subplot(2, 2, 1)
    plt.plot(n_values, n_values, 'b-', label='O(n) - Linear')
    plt.plot(n_values, n_values**2, 'g-', label='O(n²) - Quadratic')
    plt.plot(n_values, 2**n_values, 'r-', label='O(2ⁿ) - Exponential')
    plt.yscale('log')
    plt.xlabel('Input Size (n)')
    plt.ylabel('Operations (log scale)')
    plt.title('Growth Rate Comparison')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Show memoization benefit
    plt.subplot(2, 2, 2)
    stairs = list(range(1, 21))
    # Simulate relative performance (actual would be too slow for large n)
    naive_time = [3**min(n, 15) for n in stairs]  # Cap at 15 to avoid overflow
    memo_time = stairs  # Linear with memoization
    
    plt.plot(stairs, naive_time, 'r-', label='Naive Recursion O(3ⁿ)')
    plt.plot(stairs, memo_time, 'b-', label='With Memoization O(n)')
    plt.yscale('log')
    plt.xlabel('Number of Stairs')
    plt.ylabel('Relative Time (log scale)')
    plt.title('Memoization Performance Benefit')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    # Wildcard binary growth
    plt.subplot(2, 2, 3)
    wildcards = list(range(1, 11))
    results_count = [2**w for w in wildcards]
    
    plt.plot(wildcards, results_count, 'mo-', linewidth=2, markersize=6)
    plt.yscale('log')
    plt.xlabel('Number of Wildcards')
    plt.ylabel('Number of Results')
    plt.title('Wildcard Binary: O(2ʷ) Growth')
    plt.grid(True, alpha=0.3)
    
    # Factorial growth for permutations
    plt.subplot(2, 2, 4)
    n_vals = list(range(1, 8))
    factorial_vals = [np.math.factorial(n) for n in n_vals]
    
    plt.plot(n_vals, factorial_vals, 'co-', linewidth=2, markersize=6)
    plt.yscale('log')
    plt.xlabel('n (letters)')
    plt.ylabel('n! (permutations)')
    plt.title('Permutations: O(n!) Growth')
    plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    print("\nKEY TAKEAWAYS:")
    print("- Memoization can dramatically improve performance (exponential → linear)")
    print("- Exponential algorithms become impractical for large inputs")
    print("- Choose the right recursive approach based on problem constraints")
    print("- Always consider iterative alternatives for simple problems")

analyze_complexity()
````

## Summary and Key Concepts

### What We've Learned:

1. **Recursive Design Pattern**:
   - Always identify the base case first
   - Express the problem in terms of smaller subproblems
