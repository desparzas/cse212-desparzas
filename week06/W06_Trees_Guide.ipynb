## Problem 1: Insert Unique Values Only

**Goal**: Modify the Insert function to only allow unique values, creating a sorted set.

**Problem**: The original Insert function allows duplicates, which violates the sorted set property.

**Solution Strategy**:
1. Compare the new value with the current node's data
2. If value < data: go left
3. If value > data: go right  
4. If value == data: **do nothing** (ignore duplicate)

### C# Implementation:
```csharp
public void Insert(int value)
{
    if (value < Data)
    {
        // Insert to the left
        if (Left is null)
            Left = new Node(value);
        else
            Left.Insert(value);
    }
    else if (value > Data)  // Changed from 'else' to 'else if'
    {
        // Insert to the right
        if (Right is null)
            Right = new Node(value);
        else
            Right.Insert(value);
    }
    // If value == Data, do nothing (no duplicates allowed)
}
```

**Key Change**: Added the condition `else if (value > Data)` instead of just `else`. This ensures that when `value == Data`, neither branch executes, effectively ignoring the duplicate.

````python
# Let's simulate the insert unique values behavior
class SimpleBSTNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
    
    def insert_unique(self, value):
        """Insert value only if it's unique"""
        if value < self.data:
            if self.left is None:
                self.left = SimpleBSTNode(value)
                return True  # Successfully inserted
            else:
                return self.left.insert_unique(value)
        elif value > self.data:
            if self.right is None:
                self.right = SimpleBSTNode(value)
                return True  # Successfully inserted
            else:
                return self.right.insert_unique(value)
        else:
            return False  # Duplicate - not inserted
    
    def to_list(self):
        """In-order traversal to get sorted list"""
        result = []
        if self.left:
            result.extend(self.left.to_list())
        result.append(self.data)
        if self.right:
            result.extend(self.right.to_list())
        return result

# Test the unique insertion behavior
print("=== Testing Insert Unique Values ===")
root = SimpleBSTNode(15)

test_values = [10, 24, 10, 14, 33, 15, 3, 24]  # Note duplicates: 10, 15, 24
inserted_values = [15]  # Root is already inserted

for value in test_values:
    success = root.insert_unique(value)
    if success:
        inserted_values.append(value)
        print(f"✓ Inserted {value}")
    else:
        print(f"✗ Rejected duplicate {value}")

print(f"\nValues in insertion order: {inserted_values}")
print(f"Final tree (sorted): {root.to_list()}")
print(f"\nExpected result: [3, 10, 14, 15, 24, 33]")
print(f"Actual result:   {root.to_list()}")
print(f"Match: {root.to_list() == [3, 10, 14, 15, 24, 33]}")
````

## Problem 2: Contains Function Implementation

**Goal**: Implement a recursive search function to find if a value exists in the tree.

**Algorithm**:
1. **Base case**: If current value equals target, return true
2. **Recursive cases**:
   - If target < current: search left subtree
   - If target > current: search right subtree
3. **Not found**: If we reach a null node, return false

**Time Complexity**: O(log n) for balanced trees, O(n) for unbalanced

### C# Implementation:
```csharp
public bool Contains(int value)
{
    if (value == Data)
    {
        return true;  // Found it!
    }
    else if (value < Data)
    {
        // Search left subtree
        if (Left is null)
            return false;  // Not found
        else
            return Left.Contains(value);
    }
    else
    {
        // Search right subtree
        if (Right is null)
            return false;  // Not found
        else
            return Right.Contains(value);
    }
}
```

````python
# Let's trace through the Contains algorithm
class BSTWithSearch(SimpleBSTNode):
    def contains(self, value, path=[]):
        """Search for value with path tracking"""
        current_path = path + [self.data]
        
        if value == self.data:
            return True, current_path
        elif value < self.data:
            if self.left is None:
                return False, current_path + ['NULL (not found)']
            else:
                return self.left.contains(value, current_path)
        else:
            if self.right is None:
                return False, current_path + ['NULL (not found)']
            else:
                return self.right.contains(value, current_path)

# Build our test tree: 15, 10, 24, 3, 14, 33
print("=== Building Test Tree ===")
root = BSTWithSearch(15)
for value in [10, 24, 3, 14, 33]:
    root.insert_unique(value)

print(f"Tree structure: {root.to_list()}")
print("\n=== Testing Contains Function ===")

# Test cases: some exist, some don't
test_searches = [3, 14, 15, 20, 33, 1, 25]

for search_value in test_searches:
    found, path = root.contains(search_value)
    path_str = ' → '.join(map(str, path))
    status = "✓ FOUND" if found else "✗ NOT FOUND"
    print(f"Search {search_value:2d}: {status:12} | Path: {path_str}")

print("\n=== Performance Analysis ===")
print(f"Tree height: {len([15, 10, 3])} (max depth from root to leaf)")
print(f"For a balanced tree with {len(root.to_list())} nodes:")
print(f"- Expected max comparisons: {int(np.ceil(np.log2(len(root.to_list()) + 1)))}")
print(f"- Actual max comparisons observed: {max(len(root.contains(v)[1]) for v in test_searches if not root.contains(v)[0])}")
````

## Problem 3: Traverse Backwards Implementation

**Goal**: Implement reverse in-order traversal to visit nodes from largest to smallest.

**Forward Traversal (Left → Root → Right)**: Gives ascending order
**Backward Traversal (Right → Root → Left)**: Gives descending order

### Traversal Comparison:

| Traversal Type | Order | Pattern | Result for BST |
|----------------|-------|---------|----------------|
| Pre-order | Root → Left → Right | Visit root first | Not sorted |
| In-order | Left → Root → Right | Visit in sequence | Ascending sort |
| Post-order | Left → Right → Root | Visit root last | Not sorted |
| **Reverse In-order** | **Right → Root → Left** | **Reverse sequence** | **Descending sort** |

### C# Implementation:
```csharp
private void TraverseBackward(Node? node, List<int> values)
{
    if (node is not null)
    {
        TraverseBackward(node.Right, values);  // Visit right subtree first
        values.Add(node.Data);                 // Visit current node
        TraverseBackward(node.Left, values);   // Visit left subtree last
    }
}
```

````python
# Let's visualize the traversal process
class BSTWithTraversal(BSTWithSearch):
    def traverse_forward(self, result=None, depth=0):
        """In-order traversal: Left → Root → Right"""
        if result is None:
            result = []
        
        # Go left first
        if self.left:
            self.left.traverse_forward(result, depth + 1)
        
        # Visit current node
        result.append(self.data)
        print(f"{'  ' * depth}Visit: {self.data} (depth {depth})")
        
        # Go right last
        if self.right:
            self.right.traverse_forward(result, depth + 1)
        
        return result
    
    def traverse_backward(self, result=None, depth=0):
        """Reverse in-order traversal: Right → Root → Left"""
        if result is None:
            result = []
        
        # Go right first
        if self.right:
            self.right.traverse_backward(result, depth + 1)
        
        # Visit current node
        result.append(self.data)
        print(f"{'  ' * depth}Visit: {self.data} (depth {depth})")
        
        # Go left last
        if self.left:
            self.left.traverse_backward(result, depth + 1)
        
        return result

# Build test tree
print("=== Tree Structure ===")
root = BSTWithTraversal(15)
for value in [10, 24, 3, 14, 33]:
    root.insert_unique(value)

print(f"Tree: {root.to_list()}")
print("\nTree visualization:")
print("       15")
print("      /  \\")
print("    10    24")
print("   / \\     \\")
print("  3  14    33")

print("\n=== Forward Traversal (Left → Root → Right) ===")
forward_result = root.traverse_forward()
print(f"Result: {forward_result}")

print("\n=== Backward Traversal (Right → Root → Left) ===")
backward_result = root.traverse_backward()
print(f"Result: {backward_result}")

print("\n=== Comparison ===")
print(f"Forward (ascending):  {forward_result}")
print(f"Backward (descending): {backward_result}")
print(f"Is backward the reverse of forward? {backward_result == forward_result[::-1]}")
````

## Problem 4: Tree Height Calculation

**Goal**: Calculate the height of a tree using recursion.

**Definition**: Height = 1 + max(left_height, right_height)
- **Leaf node**: height = 1 (no children)
- **Internal node**: height = 1 + max height of subtrees
- **Empty tree**: height = 0

**Algorithm**:
1. If node is null, return 0
2. Recursively get left subtree height
3. Recursively get right subtree height  
4. Return 1 + max(left_height, right_height)

### C# Implementation:
```csharp
public int GetHeight()
{
    int leftHeight = 0;
    int rightHeight = 0;
    
    // Get height of left subtree
    if (Left is not null)
        leftHeight = Left.GetHeight();
        
    // Get height of right subtree
    if (Right is not null)
        rightHeight = Right.GetHeight();
        
    // Return the maximum height plus 1 for current node
    return Math.Max(leftHeight, rightHeight) + 1;
}
```

````python
# Let's visualize height calculation
class BSTWithHeight(BSTWithTraversal):
    def get_height_detailed(self, depth=0):
        """Calculate height with detailed trace"""
        indent = "  " * depth
        print(f"{indent}Calculating height for node {self.data}")
        
        left_height = 0
        right_height = 0
        
        if self.left:
            print(f"{indent}  Going left to {self.left.data}")
            left_height = self.left.get_height_detailed(depth + 1)
        else:
            print(f"{indent}  Left is null, height = 0")
            
        if self.right:
            print(f"{indent}  Going right to {self.right.data}")
            right_height = self.right.get_height_detailed(depth + 1)
        else:
            print(f"{indent}  Right is null, height = 0")
            
        height = max(left_height, right_height) + 1
        print(f"{indent}Node {self.data}: max({left_height}, {right_height}) + 1 = {height}")
        return height
    
    def get_height(self):
        """Simple height calculation"""
        left_height = self.left.get_height() if self.left else 0
        right_height = self.right.get_height() if self.right else 0
        return max(left_height, right_height) + 1

# Test with different tree structures
print("=== Height Calculation Demo ===")

# Balanced tree
print("\n1. Balanced Tree:")
balanced = BSTWithHeight(15)
for value in [10, 24, 3, 14, 33]:
    balanced.insert_unique(value)

print(f"Tree: {balanced.to_list()}")
print("Structure:")
print("       15 (level 1)")
print("      /  \\")
print("    10    24 (level 2)")
print("   / \\     \\")
print("  3  14    33 (level 3)")
height1 = balanced.get_height_detailed()
print(f"\nFinal height: {height1}")

# Unbalanced tree (linear)
print("\n\n2. Unbalanced Tree (inserted in order):")
unbalanced = BSTWithHeight(3)
for value in [10, 14, 15, 24, 33]:
    unbalanced.insert_unique(value)

print(f"Tree: {unbalanced.to_list()}")
print("Structure:")
print("3 (level 1)")
print(" \\")
print("  10 (level 2)")
print("   \\")
print("    14 (level 3)")
print("     \\")
print("      15 (level 4)")
print("       \\")
print("        24 (level 5)")
print("         \\")
print("          33 (level 6)")
height2 = unbalanced.get_height()
print(f"\nHeight: {height2}")

print("\n=== Performance Impact ===")
print(f"Balanced tree height:   {height1} (efficient: O(log n))")
print(f"Unbalanced tree height: {height2} (inefficient: O(n))")
print(f"Same data, {height2 - height1} levels difference!")
````

## Problem 5: Create Tree from Sorted List

**Goal**: Create a balanced BST from a sorted array to avoid O(n) performance.

**Problem**: Inserting sorted values sequentially creates an unbalanced tree (like a linked list).

**Solution**: Use divide-and-conquer approach:
1. Find the middle element of the range
2. Insert the middle element as root
3. Recursively process left half (elements before middle)
4. Recursively process right half (elements after middle)

**Example**: For array [10, 20, 30, 40, 50, 60]
- Insert 30 first (middle)
- Then insert 10 (middle of left half: [10, 20])
- Then insert 50 (middle of right half: [40, 50, 60])
- Continue recursively...

### C# Implementation:
```csharp
private static void InsertMiddle(int[] sortedNumbers, int first, int last, BinarySearchTree bst)
{
    // Base case: if first > last, we're done with this range
    if (first > last)
        return;
        
    // Find the middle index
    int middle = (first + last) / 2;
    
    // Insert the middle value into the BST
    bst.Insert(sortedNumbers[middle]);
    
    // Recursively insert the left half
    InsertMiddle(sortedNumbers, first, middle - 1, bst);
    
    // Recursively insert the right half  
    InsertMiddle(sortedNumbers, middle + 1, last, bst);
}
```

````python
# Let's visualize the balanced tree creation process
def insert_middle_trace(arr, first, last, result=None, depth=0):
    """Trace the insertion process for creating balanced BST"""
    if result is None:
        result = []
    
    if first > last:
        return result
    
    middle = (first + last) // 2
    indent = "  " * depth
    
    print(f"{indent}Range [{first}:{last}] -> Insert {arr[middle]} (index {middle})")
    result.append(arr[middle])
    
    # Process left half
    if first <= middle - 1:
        print(f"{indent}  Left half: [{first}:{middle-1}]")
        insert_middle_trace(arr, first, middle - 1, result, depth + 1)
    
    # Process right half
    if middle + 1 <= last:
        print(f"{indent}  Right half: [{middle+1}:{last}]")
        insert_middle_trace(arr, middle + 1, last, result, depth + 1)
    
    return result

def create_balanced_tree(arr):
    """Create balanced BST from sorted array"""
    tree = BSTWithHeight(arr[0])  # This will be overwritten
    tree = None  # Start fresh
    
    def insert_middle(first, last):
        nonlocal tree
        if first > last:
            return
        
        middle = (first + last) // 2
        if tree is None:
            tree = BSTWithHeight(arr[middle])
        else:
            tree.insert_unique(arr[middle])
        
        insert_middle(first, middle - 1)
        insert_middle(middle + 1, last)
    
    insert_middle(0, len(arr) - 1)
    return tree

def create_unbalanced_tree(arr):
    """Create unbalanced BST by inserting in order"""
    tree = BSTWithHeight(arr[0])
    for value in arr[1:]:
        tree.insert_unique(value)
    return tree

# Test with example array
test_array = [10, 20, 30, 40, 50, 60]

print("=== Creating Balanced BST from Sorted Array ===")
print(f"Input array: {test_array}")
print("\nInsertion order for balanced tree:")
insertion_order = insert_middle_trace(test_array, 0, len(test_array) - 1)
print(f"\nInsertion sequence: {insertion_order}")

# Create both trees for comparison
balanced_tree = create_balanced_tree(test_array)
unbalanced_tree = create_unbalanced_tree(test_array)

print("\n=== Comparison ===")
print(f"Balanced tree height:   {balanced_tree.get_height()}")
print(f"Unbalanced tree height: {unbalanced_tree.get_height()}")
print(f"Both contain same data: {balanced_tree.to_list() == unbalanced_tree.to_list()}")
print(f"Data: {balanced_tree.to_list()}")

# Visualize the efficiency difference
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 8))

# Performance comparison for different array sizes
sizes = np.array([7, 15, 31, 63, 127, 255, 511])
balanced_heights = np.ceil(np.log2(sizes + 1))  # Theoretical balanced height
unbalanced_heights = sizes  # Worst case unbalanced height

ax1.plot(sizes, balanced_heights, 'g-o', linewidth=3, markersize=8, label='Balanced BST - O(log n)')
ax1.plot(sizes, unbalanced_heights, 'r-o', linewidth=3, markersize=8, label='Unbalanced BST - O(n)')
ax1.set_xlabel('Number of Nodes', fontsize=12)
ax1.set_ylabel('Tree Height', fontsize=12)
ax1.set_title('Tree Height Comparison', fontsize=14, fontweight='bold')
ax1.legend(fontsize=12)
ax1.grid(True, alpha=0.3)
ax1.set_yscale('log')

# Search operations comparison
search_balanced = balanced_heights  # Max comparisons for balanced
search_unbalanced = unbalanced_heights  # Max comparisons for unbalanced

ax2.plot(sizes, search_balanced, 'g-o', linewidth=3, markersize=8, label='Balanced BST Search')
ax2.plot(sizes, search_unbalanced, 'r-o', linewidth=3, markersize=8, label='Unbalanced BST Search')
ax2.set_xlabel('Number of Nodes', fontsize=12)
ax2.set_ylabel('Max Comparisons for Search', fontsize=12)
ax2.set_title('Search Performance Comparison', fontsize=14, fontweight='bold')
ax2.legend(fontsize=12)
ax2.grid(True, alpha=0.3)
ax2.set_yscale('log')

plt.tight_layout()
plt.show()

print("\n=== Key Insights ===")
print("• Balanced BST height grows as O(log n) - very efficient")
print("• Unbalanced BST height grows as O(n) - inefficient")
print("• For 1000 nodes: Balanced ≈ 10 levels, Unbalanced = 1000 levels")
print("• Creating balanced tree from sorted data prevents worst-case performance")
````
